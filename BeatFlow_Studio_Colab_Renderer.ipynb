```python
# @title 🎬 4. Mulai Render Video Visualizer (GPU Accelerated)
# @markdown Tekan tombol Play untuk memulai rendering pixel-perfect frame demi frame.
import subprocess, time, os

os.chdir("/content/beatflow-render")

cmd = [
    "node", "scripts/cloud-render.cjs",
    "--gpu"
]

if CONFIG_FILE_PATH and os.path.exists(CONFIG_FILE_PATH):
    cmd.extend(["--config-file", CONFIG_FILE_PATH])

if AUDIO_FILE_PATH and os.path.exists(AUDIO_FILE_PATH):
    cmd.extend(["--audio-url", AUDIO_FILE_PATH])

print(f"🚀 Menjalankan Pipeline Render dengan GPU...\n")
start_time = time.time()

proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)

for line in proc.stdout:
    print(line, end="")

proc.wait()
render_duration = round(time.time() - start_time, 1)

if proc.returncode == 0:
    print(f"\n🎉 RENDER SELESAI SUKSES DALAM {render_duration} DETIK!")
else:
    print(f"\n❌ Render terhenti dengan status code: {proc.returncode}")
```

```python
# @title 🎵 3. Pilih Sumber Proyek & Audio
# @markdown Tentukan sumber konfigurasi proyek visualizer:
# @markdown - **Upload File Sendiri**: Upload file `beatflow-project.json` dan lagu `.mp3` dari PC Anda.
# @markdown - **Gunakan dari GitHub**: Mengambil data proyek & audio yang sudah pernah di-push ke repository `ditesvotreamour/render`.
import os, glob, shutil
from google.colab import files

Sumber_Proyek = "Gunakan dari GitHub (Terbaru)" # @param ["Upload File Sendiri (.json & .mp3)", "Gunakan dari GitHub (Terbaru)"]

CONFIG_FILE_PATH = ""
AUDIO_FILE_PATH = ""

if Sumber_Proyek == "Upload File Sendiri (.json & .mp3)":
    print("📤 1. Silakan upload file konfigurasi proyek BeatFlow (.json):")
    uploaded_json = files.upload()
    for fname in uploaded_json.keys():
        if fname.endswith('.json'):
            CONFIG_FILE_PATH = os.path.abspath(fname)
            print(f"✅ File JSON konfigurasi terpilih: {CONFIG_FILE_PATH}")
            break

    print("\n🎵 2. Silakan upload file audio lagu (.mp3 / .wav / .m4a):")
    uploaded_audio = files.upload()
    for fname in uploaded_audio.keys():
        if fname.lower().endswith(('.mp3', '.wav', '.m4a', '.ogg', '.flac')):
            AUDIO_FILE_PATH = os.path.abspath(fname)
            print(f"✅ File Audio terpilih: {AUDIO_FILE_PATH}")
            break
else:
    # Ambil konfigurasi terbaru dari folder render-configs/
    cfgs = sorted(glob.glob("/content/beatflow-render/render-configs/*.json"))
    if cfgs:
        CONFIG_FILE_PATH = cfgs[-1]
        print(f"✅ Menggunakan konfigurasi GitHub terbaru: {os.path.basename(CONFIG_FILE_PATH)}")
    else:
        CONFIG_FILE_PATH = "/content/beatflow-render/config.json"
        print(f"ℹ️ Menggunakan config default: {CONFIG_FILE_PATH}")

    # Ambil audio terbaru dari audio-uploads/
    auds = sorted(glob.glob("/content/beatflow-render/audio-uploads/*.*"))
    if auds:
        AUDIO_FILE_PATH = auds[-1]
        print(f"✅ Menggunakan audio GitHub terbaru: {os.path.basename(AUDIO_FILE_PATH)}")
    else:
        AUDIO_FILE_PATH = "/content/beatflow-render/public/sample-audio.mp3"
        print(f"ℹ️ Menggunakan audio sample bawaan: {AUDIO_FILE_PATH}")

print("\n🎯 Sumber siap! Silakan lanjut ke Cell 4 untuk memulai render.")
```

```python
# @title 🎬 4. Mulai Render Video Visualizer (GPU Accelerated)
# @markdown Tekan tombol Play untuk memulai rendering pixel-perfect frame demi frame.
import subprocess, time, os

os.chdir("/content/beatflow-render")

cmd = [
    "node", "scripts/cloud-render.cjs",
    "--gpu"
]

# Tambahkan CHROME_PATH jika ditemukan
if CHROME_PATH:
    cmd.extend(["--chrome-path", CHROME_PATH])

if CONFIG_FILE_PATH and os.path.exists(CONFIG_FILE_PATH):
    cmd.extend(["--config-file", CONFIG_FILE_PATH])

if AUDIO_FILE_PATH and os.path.exists(AUDIO_FILE_PATH):
    cmd.extend(["--audio-url", AUDIO_FILE_PATH])

print(f"🚀 Menjalankan Pipeline Render dengan GPU...\n")
start_time = time.time()

proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)

for line in proc.stdout:
    print(line, end="")

proc.wait()
render_duration = round(time.time() - start_time, 1)

if proc.returncode == 0:
    print(f"\n🎉 RENDER SELESAI SUKSES DALAM {render_duration} DETIK!")
else:
    print(f"\n❌ Render terhenti dengan status code: {proc.returncode}")
```

```python
# @title 📥 5. Download Video & Tonton Pratinjau
# @markdown Jalankan cell ini untuk mengunduh video MP4 hasil render ke komputer Anda.
import os, glob, shutil
from google.colab import files
from IPython.display import HTML
from base64 import b64encode

videos = sorted(glob.glob("/content/beatflow-render/dist-render/*-visualizer.mp4"))
if not videos:
    videos = sorted(glob.glob("/content/beatflow-render/dist-render/*.mp4"))

if videos:
    target_video = videos[-1]
    size_mb = round(os.path.getsize(target_video) / (1024 * 1024), 2)
    print(f"🎥 Video Ditemukan: {os.path.basename(target_video)} ({size_mb} MB)")

    # Simpan ke Google Drive jika folder terhubung
    drive_folder = "/content/drive/MyDrive/BeatFlow_Renders"
    if os.path.exists(drive_folder):
        dest_path = os.path.join(drive_folder, os.path.basename(target_video))
        shutil.copy(target_video, dest_path)
        print(f"💾 Berhasil disimpan ke Google Drive: {dest_path}")

    # Download langsung ke komputer browser
    print("\n⬇️ Memulai pengunduhan file MP4 ke komputer Anda...")
    files.download(target_video)
else:
    print("⚠️ Belum ada file video MP4 hasil render di folder dist-render/.")
```
